```python
from langchain_classic.document_loaders import DirectoryLoader,TextLoader

loader = DirectoryLoader(
    "./data/sa_doc",
    glob="*.txt",
    loader_cls=TextLoader,
    # loader_kwargs={"encoding":"utf-8"}
)

documents=loader.load()

print(type(documents))
print(documents[0].page_content)
print(documents[0].metadata)
```

```python
#chunks
from langchain_classic.text_splitter import RecursiveCharacterTextSplitter

text_splitter = RecursiveCharacterTextSplitter(
    separators=[" "],
    chunk_size=250,
    chunk_overlap=50
)


chunks = text_splitter.split_documents(documents)

print(f"no of chunks {len(chunks)} from {documents} documents")
print(f"page content preview: {chunks[0].page_content[0:100]}")
print(f"metadata preview: {chunks[0].metadata}")
```

```python
#embed + store
from langchain_openai import OpenAIEmbeddings, ChatOpenAI
from langchain_classic.vectorstores import Chroma

llm = ChatOpenAI(
    model="gpt-5.6-luna"
)

persist_directory = "./chroma_db"

vectorstore = Chroma.from_documents(
    documents=chunks,
    embedding=OpenAIEmbeddings(),
    persist_directory=persist_directory,
    collection_name="rag_collections"
)
```

```python
#retriever

retriever = vectorstore.as_retriever(
    kwargs={"k": 3}
)
```

```python
#prompt
from langchain_core.prompts import ChatPromptTemplate

system_prompt = """You are an assistant for question-answering tasks. 
Use the following pieces of retrieved context to answer the question. 
If you don't know the answer, just say that you don't know. 
Use three sentences maximum and keep the answer concise.

Context: {context}"""

prompt = ChatPromptTemplate.from_messages([
    ("system",system_prompt),
    ("human", "{input}")
])
```

```python
#document chain
from langchain_classic.chains.combine_documents import create_stuff_documents_chain

document_chain = create_stuff_documents_chain(llm,prompt)
```

```python
#retriever chain
from langchain_classic.chains.retrieval import create_retrieval_chain

rag_chain = create_retrieval_chain(retriever,document_chain)
```

```python
rag_chain.invoke({"input":"what is name of new airport opened in AP?"})
```

```python
#LCEL imports, flexible appraoch
from langchain_core.output_parsers import  StrOutputParser
from langchain_core.runnables import RunnablePassthrough, RunnableParallel
```

```python
# building custom prompt for llm
custom_prompt = ChatPromptTemplate.from_template("""You are an assistant for question-answering tasks. 
Use the following pieces of retrieved context to answer the question. 
If you don't know the answer, just say that you don't know. 
Use three sentences maximum and keep the answer concise.

Context: {context}

Question: {question}

Answer:                                              
""")
```

```python
def format_docs(docs):
    return "\n\n".join(doc.page_content for doc in docs)
```

```python
# building a LCEL rag chain

rag_chain_lcel = (
    {
        "context": retriever | format_docs,
        "question": RunnablePassthrough()
    }
    | custom_prompt 
    | llm 
    | StrOutputParser()
)
                
```

```python
rag_chain_lcel.invoke("what is latest airport build in india? which is currently running")
```
